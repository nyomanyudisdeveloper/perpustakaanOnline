```python
def penentuan_harga(x):
    total_harga = 0 
    for i in range(x):
        i = i + 1 
        if i > 0 and i <= 1000:
            total_harga = total_harga + 1500
        elif i > 1000 and i <= 5000:
            total_harga = total_harga + 2000
        else:
            total_harga = total_harga + 2500
    return total_harga
```
